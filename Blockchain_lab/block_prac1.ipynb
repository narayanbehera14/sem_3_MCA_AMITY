```python
import hashlib

def get_sha256(text: str) -> str:
    # Hash functions in hashlib require byte inputs
    return hashlib.sha256(text.encode("utf-8")).hexdigest()

# Question 1: User input and hash generation
user_input_1 = input("Enter the original text: ")
hash_1 = get_sha256(user_input_1)

print(f"\nOriginal Input:  '{user_input_1}'")
print(f"SHA-256 Hash 1:  {hash_1}")

# Question 2: Modified input with one character changed
user_input_2 = input("\nEnter modified text (change just one character): ")
hash_2 = get_sha256(user_input_2)

print(f"\nModified Input:  '{user_input_2}'")
print(f"SHA-256 Hash 2:  {hash_2}")
```

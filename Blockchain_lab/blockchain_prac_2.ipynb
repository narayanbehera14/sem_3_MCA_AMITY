```python
# 1. Take 5 inputs from the user
user_list = []
print("Enter 5 values:")
for i in range(5):
    val = input(f"Value {i + 1}: ")
    user_list.append(val)

# 2. Display all elements
print("\nElements in the list:")
for item in user_list:
    print(item, end=" ")
print()

# 3. Calculate and display hash code
# Note: Lists are mutable, so we convert to a tuple to obtain the hash code
list_hash = hash(tuple(user_list))
print(f"Hash code of the list: {list_hash}")
```

```python
import hashlib

class Block:
    def __init__(self, data, previous_hash="0"):
        self.data = data
        self.previousHash = previous_hash
        self.hash = self.calculateHash()

    def calculateHash(self):
        # Concatenate block contents to form the input string
        block_content = f"{self.data}{self.previousHash}"
        return hashlib.sha256(block_content.encode('utf-8')).hexdigest()

# Creating 3 blocks and linking them
# Block 1 (Genesis Block)
block1 = Block(data="Genesis Block", previous_hash="0")

# Block 2 (Linked to Block 1)
block2 = Block(data="Transaction: Alice -> Bob 10 BTC", previous_hash=block1.hash)

# Block 3 (Linked to Block 2)
block3 = Block(data="Transaction: Bob -> Charlie 5 BTC", previous_hash=block2.hash)

# Storing in a list to display relationships
blockchain = [block1, block2, block3]

for index, block in enumerate(blockchain, start=1):
    print(f"--- Block {index} ---")
    print(f"Block Data:     {block.data}")
    print(f"Previous Hash:  {block.previousHash}")
    print(f"Current Hash:   {block.hash}")
    print(f"Hash Length:    {len(block.hash)} characters\n")
```

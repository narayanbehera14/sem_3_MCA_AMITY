### Pixel Value Extraction
Images in OpenCV are represented as NumPy arrays. You can access pixel values using array indexing: `image[y, x]`. Note that OpenCV uses **BGR** (Blue, Green, Red) format by default.

```python
import cv2

# Get image dimensions
height, width, channels = image.shape
print(f"Image Shape: {height}x{width} with {channels} channels")

# 1. Access a specific pixel (e.g., at the center of the image)
py, px = height // 2, width // 2
pixel_value = image[py, px]

print(f"Pixel value at (y={py}, x={px}): {pixel_value}")

# 2. Extract individual channels for that pixel
blue = image[py, px, 0]
green = image[py, px, 1]
red = image[py, px, 2]

print(f"Blue: {blue}, Green: {green}, Red: {red}")

# 3. Get a region of interest (ROI) - e.g., a 5x5 block
roi = image[py:py+5, px:px+5]
print("\n5x5 Region pixel data:")
print(roi)
```

### Visualizing Individual Color Channels
OpenCV uses BGR format. Below we split the image and display each channel separately. High brightness in a channel means that color is strongly present in that area.

```python
import cv2
from google.colab.patches import cv2_imshow

# Split the image into its components
b, g, r = cv2.split(image)

print("Blue Channel:")
cv2_imshow(b)

print("Green Channel:")
cv2_imshow(g)

print("Red Channel:")
cv2_imshow(r)

# Printing shape to confirm they are 2D (single channel)
print(f"Single channel shape: {b.shape}")
```
